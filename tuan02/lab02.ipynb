```python
!!! Bài 9. Gradient Descent tìm hệ số đa thức
import numpy as np
import torch
polynomial = np.poly1d([1, 2, 3])
N = 20
X = np.random.randn(N, 1) * 5
Y = polynomial(X)
XX = np.hstack([
    X * X,
    X,
np.ones_like(X)
])
w = torch.randn(3, 1, requires_grad=True)
x = torch.tensor(XX, dtype=torch.float32)
y = torch.tensor(Y, dtype=torch.float32)
optimizer = torch.optim.NAdam([w], lr=0.01)
print("Initial coefficients:")
print(w)
for epoch in range(1000):
    optimizer.zero_grad()
    y_pred = x @ w
    mse = torch.mean(torch.square(y - y_pred))
    mse.backward()
    optimizer.step()
print("Learned coefficients:")
print(w)
```

```python
!!! Bài 10. Giải hệ phương trình bằng Autograd
import random
import torch
A = torch.tensor(random.random(), requires_grad=True)
B = torch.tensor(random.random(), requires_grad=True)
C = torch.tensor(random.random(), requires_grad=True)
D = torch.tensor(random.random(), requires_grad=True)
EPOCHS = 2000
optimizer = torch.optim.NAdam([A, B, C, D], lr=0.01)
for _ in range(EPOCHS):
    y1 = A + B - 9
    y2 = C - D - 1
    y3 = A + C - 8
    y4 = B - D - 2
    sqerr = y1*y1 + y2*y2 + y3*y3 + y4*y4
    optimizer.zero_grad()
    sqerr.backward()
    optimizer.step()
print("A =", A)
print("B =", B)
print("C =", C)
print("D =", D)
```

```python
!!! Bài 11. Viết chương trình PyTorch để thực hiện các yêu cầu sau:
import torch
import numpy as np

!!! a.) Tạo Tensor X có 24 phần tử và reshape lần lượt thành 4 × 6, 2 × 12 và 2 × 3 × 4.
# Tạo tensor X có 24 phần tử (từ 0 đến 23)
X = torch.arange(24)
print("X ban đầu (24 phần tử):", X)

# Reshape thành 4x6, 2x12 và 2x3x4
X_4x6 = X.reshape(4, 6)
X_2x12 = X.reshape(2, 12)
X_2x3x4 = X.reshape(2, 3, 4)

print("\nX reshape 4x6:\n", X_4x6)
print("\nX reshape 2x12:\n", X_2x12)
print("\nX reshape 2x3x4:\n", X_2x3x4)


!!! b.) Tạo hai Tensor A, B kích thước 3 × 4; tính A+B, A*B, A/B, trung bình theo cột và độ lệch chuẩn theo cột.
# Tạo 2 tensor A và B kích thước 3x4 ngẫu nhiên
A = torch.randn(3, 4)
B = torch.randn(3, 4)

print("Tensor A:\n", A)
print("Tensor B:\n", B)

# Các phép toán cơ bản (Element-wise)
print("\nA + B:\n", A + B)
print("A * B:\n", A * B)
print("A / B:\n", A / B)

# Tính trung bình và độ lệch chuẩn theo cột (dim=0)
# Giả sử yêu cầu tính trên Tensor A (bạn có thể áp dụng tương tự cho B)
mean_A_col = A.mean(dim=0)
std_A_col = A.std(dim=0)
print("\nTrung bình của A theo cột:\n", mean_A_col)
print("Độ lệch chuẩn của A theo cột:\n", std_A_col)


!!! c.) Cho y = 2x2 + 5x + 3. Dùng autograd tính dy/dx tại x = 2 và kiểm tra bằng phép tính thủ công.
# Khởi tạo x = 2, requires_grad=True để theo dõi đạo hàm
x = torch.tensor(2.0, requires_grad=True)

# Khai báo hàm số y = 2x^2 + 5x + 3
y = 2 * x**2 + 5 * x + 3

# Kích hoạt tính toán lan truyền ngược để tìm đạo hàm
y.backward()

# In kết quả từ Autograd
print(f"Giá trị dy/dx tại x=2 (tính bằng Autograd): {x.grad.item()}")

# Kiểm tra bằng phép tính thủ công
# y  = 2x^2 + 5x + 3
# y' = 4x + 5
# Tại x = 2 => y' = 4(2) + 5 = 13
thu_cong = 4 * 2.0 + 5
print(f"Giá trị dy/dx tại x=2 (tính thủ công): {thu_cong}")


!!! d.) Thay đa thức trong Bài 9 bằng y = 2x2 - 3x + 5; huấn luyện và cho biết các hệ số tìm được.
# 1. Tạo dữ liệu giả lập (Ground Truth)
N = 100
X_data = np.random.randn(N, 1) * 5
# Đa thức y = 2x^2 - 3x + 5
Y_data = 2 * (X_data**2) - 3 * X_data + 5

# 2. Xây dựng ma trận đặc trưng XX bao gồm [x^2, x, 1]
XX = np.hstack([
    X_data**2, 
    X_data, 
    np.ones_like(X_data)
])

# Khởi tạo trọng số w ngẫu nhiên (kích thước 3x1)
w = torch.randn(3, 1, requires_grad=True)

# Chuyển đổi dữ liệu numpy sang torch tensor
x_tensor = torch.tensor(XX, dtype=torch.float32)
y_tensor = torch.tensor(Y_data, dtype=torch.float32)

# Khởi tạo thuật toán tối ưu (sử dụng Adam hoặc NAdam với learning rate 0.05)
optimizer = torch.optim.NAdam([w], lr=0.05)

# Huấn luyện mô hình
epochs = 2000
for epoch in range(epochs):
    # Reset gradient
    optimizer.zero_grad()
    
    # Dự đoán
    y_pred = x_tensor @ w
    
    # Tính hàm mất mát (Mean Squared Error)
    loss = torch.mean(torch.square(y_tensor - y_pred))
    
    # Lan truyền ngược và cập nhật trọng số
    loss.backward()
    optimizer.step()

# In kết quả
print("Các hệ số thực tế của đa thức (a, b, c): [2.0, -3.0, 5.0]")
print("Các hệ số tìm được sau khi huấn luyện:")
print(w.detach().numpy().flatten())
```
